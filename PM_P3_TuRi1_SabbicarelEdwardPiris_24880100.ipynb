## Latihan Mandiri
**Latihan 1.** Muat dataset lain bawaan seaborn, yaitu `tips` (`sns.load_dataset('tips')`). Tampilkan `head()`, `shape`, dan `info()`.


```python
# TULIS KODE ANDA DI SINI (Latihan 1)
import pandas as pd, seaborn as sns
#dataset tips
tips = sns.load_dataset('tips')

# menampilkan 5 baris teratas
print('---- 5 baris pertama ----')
display(tips.head())

# menampilkan ukuran (baris, kolom)
print ('\n---- Dimensi (shape) ----')
print(tips.shape)

# informasi tipe data & missing value
print ('\n---- info Dataset ----')
tips.info()
```

**Latihan 2.** Pada dataset `tips`, buat histogram kolom `total_bill` dan box plot kolom `tip`. Amati sebaran dan outlier-nya.


```python
# TULIS KODE ANDA DI SINI (Latihan 2)
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Histogram kolom total_bill
plt.figure(figsize=(8, 4))
sns.histplot(data=tips, x='total_bill', kde=True, color='teal')
plt.title('Distribusi Total Bill')
plt.xlabel('Total Bill')
plt.ylabel('Frekuensi')
plt.show()

# 2. Box plot kolom tip (cek outlier)
plt.figure(figsize=(8, 3))
sns.boxplot(data=tips, x='tip', color='coral')
plt.title('Deteksi Outlier pada Tip')
plt.xlabel('Tip')
plt.show()
```

**Latihan 3.** Buat scatter plot `total_bill` (x) vs `tip` (y) pada dataset `tips`, diwarnai menurut kolom `time` (Lunch/Dinner). Apa yang dapat Anda simpulkan?


```python
# TULIS KODE ANDA DI SINI (Latihan 3)
import matplotlib.pyplot as plt
import seaborn as sns

# Muat data (berjaga-jaga jika cell dijalankan mandiri)
tips = sns.load_dataset('tips')

# Membuat scatter plot
plt.figure(figsize=(8, 5))
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='time')

plt.title('Hubungan Total Bill vs Tip (Berdasarkan Waktu)')
plt.xlabel('Total Tagihan ($)')
plt.ylabel('Besaran Tip ($)')
plt.show()
```

**Kesimpulan : Ada hubungan berbanding lurus yang jelas. makin besar total tagihan makan (total_bill), pengunjung cenderung ngasih tip yang makin besar juga. titik titiknya membentuk pola diagonal naik ke kanan atas**

**Latihan 4.** Buat heatmap korelasi untuk fitur numerik pada dataset `tips`. Pasangan fitur mana yang paling berkorelasi?


```python
# TULIS KODE ANDA DI SINI (Latihan 4)
import matplotlib.pyplot as plt
import seaborn as sns

# Muat data tips
tips = sns.load_dataset('tips')

# Ambil kolom numerik saja (total_bill, tip, size)
num_tips = tips.select_dtypes(include='number')

# Hitung korelasi & tampilkan heatmap
plt.figure(figsize=(7, 5))
sns.heatmap(num_tips.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Heatmap Korelasi Fitur Numerik (Tips)')
plt.show()
```

**Pasangan yang paling berkorelasi kuat adalah total_bill dan tip dengan nilai korelasi sekitar 0.68.**

**Artinya, faktor yang paling sejalan dan berpengaruh langsung terhadap besarnya uang tip adalah total nilai tagihan pelanggan itu sendiri, bukan seberapa banyak rombongan orang yang datang (size).**
